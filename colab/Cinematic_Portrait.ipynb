# 🎬 Cinematic Portrait — make a Reels/TikTok video from one photo

Works in any browser, **including on your phone**. Free, no API key needed.

**How to use:** tap the ▶️ button on each cell below, from top to bottom, and wait for each one to finish (a green ✓ appears).

The video is made by moving a virtual camera over your photo and adding colour grading, film grain and light leaks. **Your face is never redrawn or reshaped.**

```python
#@title 1️⃣ Set up (takes about 1 minute)
REPO_URL = "https://github.com/zeusmess/Cenima.git"  #@param {type:"string"}
BRANCH = "main"  #@param {type:"string"}

import os, subprocess, sys
os.chdir("/content")
if not os.path.isdir("/content/Cenima"):
    result = subprocess.run(["git", "clone", "-q", "--depth", "1", "-b", BRANCH, REPO_URL, "Cenima"],
                            capture_output=True, text=True)
    if result.returncode != 0:
        print(result.stderr)
        raise SystemExit("❌ Could not download the project. If your GitHub repository is private, "
                         "see 'Colab with a private repository' in the README.")
os.chdir("/content/Cenima")
subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
sys.path.insert(0, "/content/Cenima")
print("✅ Ready! Now run step 2.")
```

```python
#@title 2️⃣ Upload your portrait photo
from google.colab import files
uploaded = files.upload()
if not uploaded:
    raise SystemExit("No photo was uploaded. Run this cell again and pick a photo.")
PHOTO = list(uploaded)[-1]
print(f"✅ Got {PHOTO}. Now run step 3.")
```

```python
#@title 3️⃣ Choose a style and make the video (1–3 minutes)
LOOK = "teal_orange"  #@param ["teal_orange", "warm_film", "moody", "natural", "noir"]
MOTION = "story"  #@param ["story", "push"]
DURATION = 10  #@param {type:"slider", min:5, max:30, step:1}
TRANSITION = "leak"  #@param ["leak", "glide", "dip"]
GRAIN = 0.5  #@param {type:"slider", min:0, max:1, step:0.05}
LIGHT_LEAKS = 0.5  #@param {type:"slider", min:0, max:1, step:0.05}
BOKEH = 0.4  #@param {type:"slider", min:0, max:1, step:0.05}
FADE_OUT = True  #@param {type:"boolean"}

import os
from cinematic import Settings, render_video

OUTPUT = os.path.splitext(os.path.basename(PHOTO))[0] + "_cinematic.mp4"
settings = Settings(duration=DURATION, look=LOOK, motion=MOTION, transition=TRANSITION,
                    grain=GRAIN, light_leaks=LIGHT_LEAKS, bokeh=BOKEH, fade_out=FADE_OUT)

last = [-1]
def show(frac, msg):
    pct = int(frac * 100)
    if pct // 10 != last[0]:
        last[0] = pct // 10
        print(f"{pct:3d}%  {msg}")

result = render_video(PHOTO, OUTPUT, settings, progress=show)
print(f"✅ Done in {result.seconds:.0f} seconds: {OUTPUT}")
if not result.face_detected:
    print("ℹ️ No face was detected, so the camera assumed the face is in the upper middle.")
```

```python
#@title 4️⃣ Watch it
from IPython.display import Video
Video(OUTPUT, embed=True, width=320, html_attributes="controls playsinline loop")
```
